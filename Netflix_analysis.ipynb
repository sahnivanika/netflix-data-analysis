```python
# Convert 'date_added' to datetime objects
# Using errors='coerce' will turn any dates it can't parse into NaT (Not a Time)
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed', errors='coerce')

# Display the data type to confirm conversion
print("Data type after converting to datetime:")
print(df['date_added'].dtype)
```

```python
# Extract only the year from the 'date_added' column
# .dt.year will return the year; .dropna() is used because NaT values will result in NaN after .dt.year
df['date_added'] = df['date_added'].dt.year

# Display the data type and head to see the results
print("\nData type after extracting year:")
print(df['date_added'].dtype)
display(df.head())
```

```python
def categorize_audience(rating):
    if rating == 'TV-Y':
        return 'Children'
    elif rating in ['TV-Y7', 'TV-G']:
        return 'Anyone'
    elif rating in ['TV-MA', 'TV-14', 'TV-PG']:
        return 'Adult'
    else:
        return 'Unspecified' # Handle ratings not explicitly listed

df['Who can Watch'] = df['rating'].apply(categorize_audience)

# Display the head of the DataFrame to see the new column
display(df.head())

# Optionally, check the value counts for the new column to see the distribution
print("\nValue counts for 'Who can Watch':")
print(df['Who can Watch'].value_counts())
```

```python
# Define the threshold year
threshold_year = 2015

# Use numpy.where to categorize based on release year
df['Release Category'] = np.where(df['release_year'] >= threshold_year, 'New', 'Old')
# new 2015 onwards and old otherwise
# Display the head to see the new column
display(df.head())

# Check the value counts for the new column
print("\nValue counts for 'Release Category':")
print(df['Release Category'].value_counts())
```

```python
# Check for missing values in the specified columns
missing_values = df[['director', 'cast', 'country', 'rating']].isnull().sum()

print("Missing values count for selected columns:")
print(missing_values)
```

```python
# Fill missing values in 'director', 'cast', and 'country' with 'Unknown'
df[['director', 'cast', 'country']] = df[['director', 'cast', 'country']].fillna('Unknown')

# Find the mode (most frequent value) of the 'rating' column
# .mode()[0] is used because .mode() can return multiple values if there's a tie,
# and we typically want the first one in case of a tie.
most_frequent_rating = df['rating'].mode()[0]

# Fill missing values in 'rating' with the most frequent rating
df['rating'] = df['rating'].fillna(most_frequent_rating)

# Verify that there are no more missing values in these columns
print("Missing values count after filling:")
print(df[['director', 'cast', 'country', 'rating']].isnull().sum())

# Display the head to see the changes (especially for columns that had missing values in the first few rows)
display(df.head())
```

```python
from pandasql import sqldf

# This query counts the number of Movies and TV Shows added by country after the year 2015
query = """
SELECT
    country,
    type,
    COUNT(*) AS content_count
FROM
    df
WHERE
    date_added > 2015  -- Filter for content added after 2015 (comparing as float)
GROUP BY
    country,
    type
ORDER BY
    content_count DESC  -- Order by the count in descending order
LIMIT 10;  -- Get the top 10 combinations
"""

# Run the SQL query on the pandas DataFrame
complex_sql_result = sqldf(query, globals())

# Display the result
print("Result of complex SQL query on pandas DataFrame (Top Countries by Content Type added after 2015):")
display(complex_sql_result)
```
